```python
import cv2
import numpy as np
```

```python
img=cv2.imread(r"C:\Users\Dell Latitude E5470\Pictures\smarties.png",0)
```

```python
ret,mask=cv2.threshold(img,220,255,cv2.THRESH_BINARY_INV)
```

```python
kernel=np.ones((5,5),np.uint8)
diltion=cv2.dilate(mask,kernel,iterations=2)
erosion=cv2.erode(mask,kernel,iterations=2)
opening=cv2.morphologyEx(mask,cv2.MORPH_OPEN,kernel,iterations=2)
closing=cv2.morphologyEx(mask,cv2.MORPH_CLOSE,kernel,iterations=2)
```

```python
#cv2.imshow("image",img)
cv2.imshow("thres",mask)
cv2.imshow("dilation",diltion)
cv2.imshow("erosion",erosion)
cv2.imshow("opening",opening)
cv2.imshow("closing",closing)
cv2.waitKey(0)
```
