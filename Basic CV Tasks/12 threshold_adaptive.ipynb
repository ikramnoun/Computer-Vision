```python
import cv2
import numpy as np
import os
```

```python
os.chdir(r"C:\computer vision")
#img=cv2.imread(r"C:\Users\Dell Latitude E5470\Pictures\gradient.png",1)
img=cv2.imread(r"C:\Users\Dell Latitude E5470\Pictures\sudoku.png",0)
print(img.shape)
```

```python

ret,thr1=cv2.threshold(img,55,255,cv2.THRESH_BINARY)
ret,thr2=cv2.threshold(img,55,255,cv2.THRESH_BINARY_INV)
ret,thr3=cv2.threshold(img,55,255,cv2.THRESH_TRUNC)
ret,thr4=cv2.threshold(img,55,255,cv2.THRESH_TOZERO)
ret,thr5=cv2.threshold(img,55,255,cv2.THRESH_TOZERO_INV)
```

```python
thr6=cv2.adaptiveThreshold(img,155,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,111,2)
thr7=cv2.adaptiveThreshold(img,155,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,111,2)
```
