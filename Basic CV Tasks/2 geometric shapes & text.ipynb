```python
import numpy as np
import cv2
```

```python
mask=np.zeros([480,640,3],np.uint8)
cv2.imshow("mask",mask)
```

```python
#line
#line=cv2.line(mask,(0,0),(320,240),(0,0,255),10)
line=cv2.arrowedLine(mask,(0,0),(320,240),(0,0,255),10)
cv2.imshow("arrowedline",line)
```

```python
rect=cv2.rectangle(mask,(0,0),(320,240),(0,255,0),10)
cv2.imshow("rectangle",rect)
```

```python
circle=cv2.circle(mask,(320,240),100,(0,255,0),10)
cv2.imshow("circle",circle)
```

```python
font=cv2.FONT_HERSHEY_SIMPLEX
text=cv2.putText(mask,'noik',(100,100),font,2,(0,255,0),10)
cv2.imshow("text",text)
```

```python
pts=np.array([[10,5],[20,30],[70,20],[50,10]],np.int32)
polylines=cv2.polylines(mask,[pts],True,(0,255,0))
cv2.imshow("polylines",polylines)
```
