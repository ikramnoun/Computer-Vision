```python
def nothing(x):
    print(x)
cv2.namedWindow('tracking')
#cv2.resizeWindow('tracking', 250, 250)
cv2.createTrackbar('H','tracking',0,179,nothing)
cv2.createTrackbar('S','tracking',0,255,nothing)
cv2.createTrackbar('V','tracking',0,255,nothing)

hsv=np.zeros((250,500,3),np.uint8)
while True:
     h=cv2.getTrackbarPos('H','tracking')    
     s=cv2.getTrackbarPos('S','tracking')  
     v=cv2.getTrackbarPos('V','tracking')

     hsv[:]=(h,s,v)
     bgr=cv2.cvtColor(hsv,cv2.COLOR_HSV2BGR)

     cv2.imshow("bgr",bgr)
     if cv2.waitKey(1)==ord("q"):
          break

```
