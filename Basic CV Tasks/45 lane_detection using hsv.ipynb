```python
import cv2
import numpy as np
```

```python
frame=cv2.imread(r"C:\Users\Dell Latitude E5470\Pictures\solidYellowCurve.jpg",1)
hsv=cv2.cvtColor(frame,cv2.COLOR_BGR2HSV)

l_y=np.array([18,94,140])
u_y=np.array([48,255,255])

mask=cv2.inRange(hsv,l_y,u_y)
edges=cv2.Canny(mask,75,150)

lines=cv2.HoughLinesP(edges,1,np.pi/180,60,maxLineGap=20)
if lines is not None:
    for line in lines:
        x1,y1,x2,y2=line[0]
        cv2.line(frame,(x1,y1),(x2,y2),(0,255,0),10)
cv2.imshow("mask",mask)
cv2.imshow("edges",edges)
cv2.imshow("frame",frame)
cv2.waitKey(0)
```
