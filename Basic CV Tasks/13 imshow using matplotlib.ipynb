```python
import cv2
import numpy
from matplotlib import pyplot as plt
```

```python
img=cv2.imread(r"C:\Users\Dell Latitude E5470\Pictures\apple.jpg",-1)
cv2.imshow('img',img)
cv2.waitKey(0)
```

```python
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.xticks([]),plt.yticks([])
plt.show()
```

```python
img2=cv2.imread(r"C:\Users\Dell Latitude E5470\Pictures\gradient.png",0)
ret,ther1=cv2.threshold(img2,127,255,cv2.THRESH_BINARY)
ret,ther2=cv2.threshold(img2,127,255,cv2.THRESH_BINARY_INV)
ret,ther3=cv2.threshold(img2,127,255,cv2.THRESH_TRUNC)
ret,ther4=cv2.threshold(img2,127,255,cv2.THRESH_TOZERO)
ret,ther5=cv2.threshold(img2,127,255,cv2.THRESH_TOZERO_INV)

titels=["original","binary","binary_inv","trunc","to_zeros","to_zeros_inv"]
imgs=[img2,ther1,ther2,ther3,ther4,ther5]

for i in range(6):
    plt.subplot(2, 3, i+1),plt.imshow(imgs[i],'gray')
    plt.title(titels[i]) 
    plt.xticks([]),plt.yticks([])
    plt.show()
```
