```python
import cv2
import numpy as np
```

```python
cap=cv2.VideoCapture(r"C:\Users\Dell Latitude E5470\Downloads\vtest.avi",0)
width=int(cap.get(3))
height=int(cap.get(4))
```

```python
ret,frame1=cap.read()
ret,frame2=cap.read()
```

```python
while True:
    if ret==True:
     diff=cv2.absdiff(frame1,frame2)
     gray=cv2.cvtColor(diff,cv2.COLOR_BGR2GRAY)
     blur=cv2.GaussianBlur(gray,(5,5),0)
     ret,thresh=cv2.threshold(blur,20,255,cv2.THRESH_BINARY)
     dilation=cv2.dilate(thresh,None,iterations=3)
     contours,_=cv2.findContours(dilation,cv2.RETR_TREE,cv2.CHAIN_APPROX_NONE)
     draw=cv2.drawContours(frame1,contours,-1,(255,0,0),2)
     for contour in contours:
        (x,y,w,h)=cv2.boundingRect(contour)
        if cv2.contourArea(contour)<900:
            continue
        cv2.putText(frame1,"state:{}".format("movement"),(10,20),cv2.FONT_HERSHEY_SIMPLEX,1,(0,255,0),3)
        cv2.rectangle(frame1,(x,y),(x+w,h+y),(0,255,0),2)
    
     cv2.imshow('motion detection',frame1)
     #cv2.imshow('threshold',thresh)
     #cv2.imshow('dilation',dilation)
     #cv2.imshow('contours',draw)
     frame1=frame2
     ret,frame2=cap.read()
     if cv2.waitKey(60)==ord("q"):
        break
    else:
       break
cv2.destroyAllWindows()
cap.release()

```
